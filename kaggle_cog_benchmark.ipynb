# Cog — Broad Validation Benchmark (Kaggle)

Self-contained run of the Cog `policy_dep_aware_prioritization` broad-validation campaign.
Uses the mounted **Cog dataset** when present (fast, no clone), otherwise clones the
latest repo from GitHub. Writes results to `/kaggle/working/results.json`
(downloadable when the notebook finishes).

**Rerun:** just click *Run All*. If the dataset is attached, it reuses the mounted code;
otherwise it re-clones the latest GitHub code.

Repo: https://github.com/aungnuepruemarma-ship-it/cog  
Dataset: https://www.kaggle.com/datasets/aungmarma/cog-runtime-source

```python
import os, glob

# Prefer a mounted Cog dataset (fast, no network). Kaggle mounts datasets under
# /kaggle/input/<slug>, and the slug may differ from the dataset id, so detect it
# by looking for the cog package + run.py launcher inside any input dir.
COG_DIR = None
for cand in glob.glob('/kaggle/input/*'):
    if os.path.isfile(os.path.join(cand, 'cog', '__init__.py')) and \
       os.path.isfile(os.path.join(cand, 'run.py')):
        COG_DIR = cand
        break

if COG_DIR:
    print('Using mounted Cog dataset at', COG_DIR)
else:
    !git clone https://github.com/aungnuepruemarma-ship-it/cog.git
    COG_DIR = 'cog'
    print('Cloned Cog repo into', COG_DIR)
%cd {COG_DIR}
```

```python
# Inspect the results Kaggle will let you download from /kaggle/working/
import json
from pathlib import Path

p = Path('/kaggle/working/results.json')
if not p.exists():
    p = Path('results.json')
print('results at:', p)
data = json.loads(p.read_text())
print('ACTIVE GO/NO-GO :', 'GO' if data['active_go'] else 'NO-GO')
print('overall delta   :', data['delta'], 'CI95', data['delta_ci95'])
print('no-regression   :', data['no_regression'])
print('rollback clean  :', data['rollback_clean'])
```
