# Cog — Controlled Rollout + Regression (Kaggle)

EXP-ROLLOUT-001: controlled activation of the dependency-aware ordering heuristic,
followed by a multi-domain regression battery. The policy is created in the
EXPERIMENTAL state (benchmark-suite-only rollout) and **not** auto-promoted —
general adoption stays a manual step.

Uses the mounted **Cog dataset** when present (fast), otherwise clones the latest
repo from GitHub. Writes results to `/kaggle/working/results.json`.

Repo: https://github.com/aungnuepruemarma-ship-it/cog  
Dataset: https://www.kaggle.com/datasets/aungmarma/cog-runtime-source

```python
import os, glob

# Prefer a mounted Cog dataset (fast, no network); fall back to git clone.
COG_DIR = None
for cand in glob.glob('/kaggle/input/*'):
    if os.path.isfile(os.path.join(cand, 'cog', '__init__.py')) and \
       os.path.isfile(os.path.join(cand, 'run.py')):
        COG_DIR = cand
        break

if COG_DIR:
    print('Using mounted Cog dataset at', COG_DIR)
else:
    !git clone https://github.com/aungnuepruemarma-ship-it/cog.git
    COG_DIR = 'cog'
    print('Cloned Cog repo into', COG_DIR)
%cd {COG_DIR}
```
